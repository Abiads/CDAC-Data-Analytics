## The 4 Pillars of Descriptive Statistics

1. **Measures of Central Tendency (Location):**
   - **Arithmetic Mean ($\bar{x}$):** Center of gravity $\frac{1}{n} \sum x_i$ (sensitive to outliers).
   - **Geometric Mean ($GM$):** $(\prod x_i)^{1/n}$ (compounding growth rates, investments).
   - **Harmonic Mean ($HM$):** $\frac{n}{\sum \frac{1}{x_i}}$ (rates, speeds, F1-scores).
   - **Median ($Q_2$):** 50th percentile, completely robust against extreme values.
   - **Mode:** Most frequent observation (useful for categorical/nominal data).

2. **Measures of Dispersion (Spread & Variability):**
   - **Range:** $\text{Max} - \text{Min}$ (crude spread).
   - **Interquartile Range (IQR):** $Q_3 - Q_1$ (middle 50% spread, robust).
   - **Sample Variance ($s^2$):** $\frac{1}{n-1} \sum (x_i - \bar{x})^2$ (`ddof=1`, Bessel's correction).
   - **Standard Deviation ($s$):** $\sqrt{s^2}$ (in original measurement units).
   - **Coefficient of Variation (CV):** $\frac{s}{\bar{x}} \times 100\%$ (dimensionless unit-free relative variability).

3. **Measures of Shape (Higher Order Moments):**
   - **Skewness ($3^{\text{rd}}$ moment):** Asymmetry ("The Mean Chases the Tail").
   - **Kurtosis ($4^{\text{th}}$ moment):** Tailedness / outlier likelihood (Leptokurtic vs Mesokurtic vs Platykurtic).

4. **Measures of Relative Standing:**
   - **Z-Score:** $z = \frac{x - \bar{x}}{s}$ (distance from mean in standard deviation units).
   - **Tukey's Fences:** $[Q_1 - 1.5 \times \text{IQR}, Q_3 + 1.5 \times \text{IQR}]$ for outlier detection.

```python
import numpy as np
import pandas as pd
from scipy import stats

# 1. When to use Geometric Mean (Compounding Investment Returns)
# Year 1: +50% (factor 1.5), Year 2: -50% (factor 0.5)
growth_factors = np.array([1.50, 0.50])
print(f"Arithmetic Mean Factor : {np.mean(growth_factors):.4f} (Misleading: suggests 0% break-even)")
print(f"Geometric Mean Factor  : {stats.gmean(growth_factors):.4f} (Actual return: -13.4% annual drag)")

# 2. When to use Harmonic Mean (Average Speed over Equal Distances)
# 60 km at 30 km/h, 60 km at 60 km/h
speeds = np.array([30.0, 60.0])
print(f"\nArithmetic Mean Speed : {np.mean(speeds):.2f} km/h (Incorrect)")
print(f"Harmonic Mean Speed   : {stats.hmean(speeds):.2f} km/h (Correct: Total Distance / Total Time)")
```

```python
# 3. Coefficient of Variation (CV) for Relative Risk
stock_A = np.array([2000, 2050, 1980, 2020, 2010])  # Blue-chip ($2000)
stock_B = np.array([20, 25, 18, 22, 21])             # Penny stock ($20)

cv_A = (np.std(stock_A, ddof=1) / np.mean(stock_A)) * 100
cv_B = (np.std(stock_B, ddof=1) / np.mean(stock_B)) * 100

print(f"Stock A (Blue Chip) : Mean={np.mean(stock_A):.1f}, Std={np.std(stock_A, ddof=1):.2f} => CV = {cv_A:.2f}%")
print(f"Stock B (Penny)     : Mean={np.mean(stock_B):.1f}, Std={np.std(stock_B, ddof=1):.2f} => CV = {cv_B:.2f}%")
print(f"Observation: Stock B has {cv_B / cv_A:.1f}x higher relative volatility despite lower raw standard deviation!")
```

```python
# 4. Outlier Detection: Z-Score vs. Tukey's 1.5 * IQR Rule
dataset = np.array([45, 52, 54, 55, 58, 60, 62, 63, 65, 68, 70, 115])

mean_val = np.mean(dataset)
std_val = np.std(dataset, ddof=1)
z_scores = (dataset - mean_val) / std_val

q1 = np.percentile(dataset, 25)
q3 = np.percentile(dataset, 75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

print(f"Dataset: {dataset.tolist()}")
print(f"Tukey's Outliers (outside [{lower_fence:.1f}, {upper_fence:.1f}]): {dataset[(dataset < lower_fence) | (dataset > upper_fence)].tolist()}")
print(f"Z-Score Outliers (|z| > 2.5)                             : {dataset[np.abs(z_scores) > 2.5].tolist()}")
```
