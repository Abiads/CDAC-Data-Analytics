### standard deviation and variance

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
incomes = np.random.normal(100,20.0,10000)
plt.hist(incomes, bins=50)
plt.show()
```

$$\sigma = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (x_i - \mu)^2}$$

```python
# what is standard deviation? 
# Standard deviation is a measure of the amount of variation or dispersion in a set of values. 

# steps for std
# 1. Calculate the mean of the data set.
# 2. Subtract the mean from each data point and square the result.
# 3. Calculate the average of these squared differences.
# 4. Take the square root of this average to get the standard deviation.
from scipy import stats

print("Income std ",incomes.std())
# using stats
import statistics
print("Income std using statistics module ", statistics.stdev(incomes))
# use scipy
print("Income std using scipy module ", stats.tstd(incomes))
```

### variance 

```python
from scipy import stats
import statistics
# what is variance?
# Variance is a measure of how much the values in a data set differ from the mean of the data set. 

# steps for variance
# 1. Calculate the mean of the data set.
# 2. Subtract the mean from each data point and square the result.
# 3. Calculate the average of these squared differences.

print("Income variance ",incomes.var())
# using stats
print("Income variance using statistics module ", statistics.variance(incomes))
# use scipy
print("Income variance using scipy module ", stats.tvar(incomes))
```

```python
data = [1,4,5,4,8]
print("original data:", data)
# step 1 calculate the mean
mean = sum(data) / len(data)
print("mean:", mean)
# step 2 find the difference from the data
differences = [x - mean for x in data]
print("differences from mean:", differences)
# step 3 square the differences
squared_differences = [d ** 2 for d in differences]
print("squared differences:", squared_differences)
# step 4 calculate the variance
variance = sum(squared_differences) / len(data)
print("variance:", variance)
# using statistics module
variance_stats = statistics.variance(data)
print("variance using statistics module:", variance_stats)

# using scipy module
variance_scipy = stats.tvar(data)
print("variance using scipy module:", variance_scipy)   

# using numpy 

variance_numpy = np.var(data)
print("variance using numpy module:", variance_numpy)

```

```python
# standard deviation
print("Income standard deviation ",incomes.std())
# using statistics module
print("Income standard deviation using statistics module ", statistics.stdev(incomes))
# using scipy module
print("Income standard deviation using scipy module ", stats.tstd(incomes))
# using numpy module
print("Income standard deviation using numpy module ", np.std(incomes))
```
