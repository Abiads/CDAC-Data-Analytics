$$P(A) = \frac{n(A)}{n(S)} = \frac{\text{Number of Favorable Outcomes}}{\text{Total Number of Possible Outcomes}}$$

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom
import matplotlib.ticker as mtick

# Parameters
n = 10  # 10 rolls
p = 1 / 6  # Probability of rolling a 6 on a single die

# All possible number of 6s you could roll (0 to 10)
x = np.arange(0, n + 1)

# Calculate probability for each outcome
probabilities = binom.pmf(x, n, p)

# Plot as a bar chart (ideal for discrete outcomes)
plt.figure(figsize=(9, 5))
bars = plt.bar(x, probabilities, color="skyblue", edgecolor="black")

# --- Add Labels on Top of Each Bar ---
# Option A: Formatted as Percentage (matches y-axis)
labels = [f"{val * 100:.1f}%" if val >= 0.0005 else "<0.1%" for val in probabilities]
plt.bar_label(bars, labels=labels, padding=3, fontsize=9)

# (Alternative Option B: If you prefer decimal values like 0.323)
# plt.bar_label(bars, fmt="%.3f", padding=3, fontsize=9)

plt.title("Probability of Rolling a '6' in 10 Die Rolls", fontsize=13, fontweight="bold")
plt.xlabel("Number of 6s Rolled (x)", fontsize=11)
plt.ylabel("Probability", fontsize=11)
plt.xticks(x)

# Give 15% headroom at the top so labels don't get cut off by the border
plt.ylim(0, max(probabilities) * 1.15)

plt.grid(axis="y", alpha=0.3)
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom

# Setup: 3 flips, 50% chance of heads
n = 3
p = 0.5
x = np.arange(0, n + 1)  # Outcomes: 0, 1, 2, or 3 heads

# Calculate probabilities
probabilities = binom.pmf(x, n, p)

# Plot as a bar chart
plt.figure(figsize=(7, 4))
bars = plt.bar(x, probabilities,  edgecolor="black")


plt.title("Probability of Getting 'x' Heads in 3 Coin Flips")
plt.xlabel("Number of Heads (x)")
plt.ylabel("Probability")
plt.xticks(x)
plt.ylim(0, 0.45)
plt.grid(axis="y", alpha=0.3)
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))
plt.show()
```
