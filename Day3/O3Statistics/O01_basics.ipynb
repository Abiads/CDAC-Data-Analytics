### Generate Random Data 
- Amount spent per transaction

```python
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(loc=100.0, scale=20.0, size=1000)
plt.hist(incomes, bins=50)
plt.show()
```

```python
print("mean:", np.mean(incomes),sep="\t:\t")
print("median:", np.median(incomes),sep="\t:\t")
```

```python
new_incomes = np.random.normal(loc=27000, scale=15000.0, size=1000) 
plt.hist(new_incomes, bins=50)
plt.show()


```

```python
# new_incomes statistics
print("mean:", np.mean(new_incomes),sep="\t:\t")
print("median:", np.median(new_incomes),sep="\t:\t")
```

```python
# add income to new_incomes
new_incomes = np.append(new_incomes, [1000000000000000])
# new_incomes statistics after adding extreme value
print("mean:", np.mean(new_incomes),sep="\t:\t")
print("median:", np.median(new_incomes),sep="\t:\t")
```

## Mode 
### Generate fake age data for 500 people

```python
ages = np.random.randint(low=18, high=90, size=500)
print("ages:", ages, sep="\t:\t")
```

```python
from scipy import stats
print("mode of ages:", stats.mode(ages), sep="\t:\t")
# most common age(s) using Counter
age_counts = Counter(ages)
print("most common age(s):", age_counts.most_common(1), sep="\t:\t")
```
