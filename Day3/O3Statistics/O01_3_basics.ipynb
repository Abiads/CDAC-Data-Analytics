## Data Distribution

```python
# uniform distribution
import numpy as np
import matplotlib.pyplot as plt
```

```python
values = np.random.uniform(low=-10.0, high=10.0, size=100000)
plt.hist(values, bins=50)
plt.show()
```

```python
from scipy.stats import norm

x = np.arange(1,5,1)
y = norm.pdf(x) # probability density function of the standard normal distribution
print("x values: ", x)
print("y values: ", y)
plt.plot(x, y)
plt.show()
```
