## 1. Step-by-Step Numerical Example (Hand Calculation)

### Problem Scenario:
A study evaluates $n = 5$ students, tracking weekly **Study Hours ($X$)** and their **Final Exam Score ($Y$)**:

| Student | Study Hours ($X$) | Exam Score ($Y$) |
| :---: | :---: | :---: |
| **A** | 2 | 50 |
| **B** | 4 | 65 |
| **C** | 6 | 60 |
| **D** | 8 | 80 |
| **E** | 10 | 95 |

---

### Step 1: Compute Sample Means
$$\bar{X} = \frac{\sum X_i}{n} = \frac{2 + 4 + 6 + 8 + 10}{5} = \frac{30}{5} = 6.0 \text{ hours}$$
$$\bar{Y} = \frac{\sum Y_i}{n} = \frac{50 + 65 + 60 + 80 + 95}{5} = \frac{350}{5} = 70.0 \text{ marks}$$

---

### Step 2: Tabular Calculation Breakdown

| Student | $X_i$ | $Y_i$ | $(X_i - \bar{X})$ | $(Y_i - \bar{Y})$ | $(X_i - \bar{X})^2$ | $(Y_i - \bar{Y})^2$ | $(X_i - \bar{X})(Y_i - \bar{Y})$ | $\text{Rank}(X)$ | $\text{Rank}(Y)$ | $d_i = R_X - R_Y$ | $d_i^2$ |
| :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: | :---: |
| **A** | 2 | 50 | $-4$ | $-20$ | $16$ | $400$ | $+80$ | 1 | 1 | $0$ | $0$ |
| **B** | 4 | 65 | $-2$ | $-5$ | $4$ | $25$ | $+10$ | 2 | 3 | $-1$ | $1$ |
| **C** | 6 | 60 | $0$ | $-10$ | $0$ | $100$ | $0$ | 3 | 2 | $+1$ | $1$ |
| **D** | 8 | 80 | $+2$ | $+10$ | $4$ | $100$ | $+20$ | 4 | 4 | $0$ | $0$ |
| **E** | 10 | 95 | $+4$ | $+25$ | $16$ | $625$ | $+100$ | 5 | 5 | $0$ | $0$ |
| **Sum ($\Sigma$)** | **30** | **350** | **0** | **0** | **40** | **1250** | **210** | — | — | **0** | **2** |

---

### Step 3: Compute Sample Variances and Standard Deviations (with Bessel's correction $n - 1 = 4$)
- **Variance of $X$ ($s_X^2$):**
  $$s_X^2 = \frac{\sum (X_i - \bar{X})^2}{n - 1} = \frac{40}{4} = 10.0 \implies s_X = \sqrt{10.0} \approx 3.1623 \text{ hours}$$
- **Variance of $Y$ ($s_Y^2$):**
  $$s_Y^2 = \frac{\sum (Y_i - \bar{Y})^2}{n - 1} = \frac{1250}{4} = 312.5 \implies s_Y = \sqrt{312.5} \approx 17.6777 \text{ marks}$$

---

### Step 4: Compute Sample Covariance ($Cov(X, Y)$)
$$Cov(X, Y) = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{n - 1} = \frac{210}{4} = 52.5 \text{ (hours} \cdot \text{marks)}$$
> **Interpretation:** The positive sign indicates that study hours and exam scores move in the same direction (students studying longer tend to score higher).

---

### Step 5: Compute Pearson Correlation Coefficient ($r$)
$$r = \frac{Cov(X, Y)}{s_X \cdot s_Y} = \frac{52.5}{3.1623 \times 17.6777} = \frac{52.5}{55.9017} \approx 0.9391$$
- **Coefficient of Determination ($R^2$):**
  $$R^2 = r^2 = (0.9391)^2 = 0.8820 \quad (88.2\% \text{ of the variance in exam scores is explained by study hours})$$

---

### Step 6: Compute Spearman Rank Correlation ($\rho$)
$$\rho = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)} = 1 - \frac{6(2)}{5(5^2 - 1)} = 1 - \frac{12}{5(24)} = 1 - \frac{12}{120} = 1 - 0.10 = 0.9000$$
> **Interpretation:** High monotonic agreement. Even though student B scored slightly higher than student C despite studying 2 fewer hours ($d_i = \pm 1$), the general ordering remains strongly monotonic.

---

### Step 7: Bivariate Ordinary Least Squares (OLS) Linear Regression
- **Slope ($\beta_1$):**
  $$\beta_1 = \frac{Cov(X, Y)}{s_X^2} = \frac{52.5}{10.0} = 5.25 \text{ marks per hour}$$
- **Intercept ($\beta_0$):**
  $$\beta_0 = \bar{Y} - \beta_1 \bar{X} = 70.0 - (5.25 \times 6.0) = 70.0 - 31.5 = 38.5 \text{ marks}$$
- **Fitted Regression Model:**
  $$\hat{Y} = 38.5 + 5.25 X$$
- **Prediction Example:** For a student studying $X = 7$ hours:
  $$\hat{Y} = 38.5 + 5.25(7) = 38.5 + 36.75 = 75.25 \text{ marks}$$

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# ==============================================================================
# NUMERICAL EXAMPLE: STUDY HOURS (X) VS EXAM SCORES (Y)
# ==============================================================================
students = ['Student A', 'Student B', 'Student C', 'Student D', 'Student E']
x = np.array([2, 4, 6, 8, 10], dtype=float)
y = np.array([50, 65, 60, 80, 95], dtype=float)
n = len(x)

# 1. Step-by-Step Arithmetic Calculation
mean_x = np.sum(x) / n
mean_y = np.sum(y) / n

dev_x = x - mean_x
dev_y = y - mean_y
sq_dev_x = dev_x ** 2
sq_dev_y = dev_y ** 2
cross_prod = dev_x * dev_y

var_x = np.sum(sq_dev_x) / (n - 1)
var_y = np.sum(sq_dev_y) / (n - 1)
std_x = np.sqrt(var_x)
std_y = np.sqrt(var_y)

cov_manual = np.sum(cross_prod) / (n - 1)
r_pearson_manual = cov_manual / (std_x * std_y)

# Spearman Rank Calculation
rank_x = stats.rankdata(x)
rank_y = stats.rankdata(y)
d = rank_x - rank_y
d_squared = d ** 2
rho_spearman_manual = 1.0 - (6.0 * np.sum(d_squared)) / (n * (n**2 - 1))

# OLS Regression Parameters
beta_1 = cov_manual / var_x
beta_0 = mean_y - beta_1 * mean_x

# Format Tabular Breakdown in Pandas
df_breakdown = pd.DataFrame({
    'Student': students,
    'X (Hours)': x,
    'Y (Score)': y,
    '(X - X̄)': dev_x,
    '(Y - Ȳ)': dev_y,
    '(X - X̄)²': sq_dev_x,
    '(Y - Ȳ)²': sq_dev_y,
    '(X - X̄)(Y - Ȳ)': cross_prod,
    'Rank(X)': rank_x.astype(int),
    'Rank(Y)': rank_y.astype(int),
    'd = Rx - Ry': d.astype(int),
    'd²': d_squared.astype(int)
})

# Add Summation Row
df_sum = pd.DataFrame([{
    'Student': 'Total Sum (Σ)',
    'X (Hours)': np.sum(x),
    'Y (Score)': np.sum(y),
    '(X - X̄)': np.sum(dev_x),
    '(Y - Ȳ)': np.sum(dev_y),
    '(X - X̄)²': np.sum(sq_dev_x),
    '(Y - Ȳ)²': np.sum(sq_dev_y),
    '(X - X̄)(Y - Ȳ)': np.sum(cross_prod),
    'Rank(X)': '',
    'Rank(Y)': '',
    'd = Rx - Ry': np.sum(d).astype(int),
    'd²': np.sum(d_squared).astype(int)
}])
df_full_table = pd.concat([df_breakdown, df_sum], ignore_index=True)

print("=" * 78)
print("           STEP-BY-STEP NUMERICAL CALCULATION TABLE (HAND AUDIT)           ")
print("=" * 78)
print(df_full_table.to_string(index=False))
print("=" * 78)

# Verification against NumPy & SciPy
cov_np = np.cov(x, y, ddof=1)[0, 1]
r_scipy, p_p = stats.pearsonr(x, y)
rho_scipy, p_s = stats.spearmanr(x, y)

print("\n--- VERIFICATION AUDIT (Manual vs NumPy/SciPy) ---")
print(f"Sample Mean X (X̄)             : {mean_x:.2f} hours")
print(f"Sample Mean Y (Ȳ)             : {mean_y:.2f} marks")
print(f"Sample Variance X (s_X²)       : {var_x:.2f} (Std Dev s_X: {std_x:.4f})")
print(f"Sample Variance Y (s_Y²)       : {var_y:.2f} (Std Dev s_Y: {std_y:.4f})")
print(f"Sample Covariance Cov(X, Y)    : Manual = {cov_manual:.4f} | NumPy = {cov_np:.4f} (Match: {np.isclose(cov_manual, cov_np)})")
print(f"Pearson Correlation (r)        : Manual = {r_pearson_manual:.4f} | SciPy = {r_scipy:.4f} (Match: {np.isclose(r_pearson_manual, r_scipy)})")
print(f"Spearman Rank Corr (rho)       : Manual = {rho_spearman_manual:.4f} | SciPy = {rho_scipy:.4f} (Match: {np.isclose(rho_spearman_manual, rho_scipy)})")
print(f"OLS Linear Equation            : Ŷ = {beta_0:.2f} + {beta_1:.2f} * X")
print(f"Coefficient of Det (R²)        : {r_pearson_manual**2:.4f} ({r_pearson_manual**2 * 100:.2f}% explained variance)")
print(f"Prediction for X = 7 Hours     : Ŷ = {beta_0 + beta_1 * 7.0:.2f} marks")
print("=" * 78)
```

```python
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

# ------------------------------------------------------------------------------
# Subplot 1: Scatter with Mean Quadrants & Fitted Regression Line
# ------------------------------------------------------------------------------
ax1.axvline(mean_x, color='gray', linestyle='--', linewidth=1.5, alpha=0.8, label=f'Mean X̄ = {mean_x:.1f}')
ax1.axhline(mean_y, color='gray', linestyle='--', linewidth=1.5, alpha=0.8, label=f'Mean Ȳ = {mean_y:.1f}')

# Label Covariance Quadrants
ax1.text(8.0, 91, 'Quadrant I\n(+)(+) = Positive', fontsize=8.5, color='darkgreen', alpha=0.75, weight='bold')
ax1.text(2.2, 91, 'Quadrant II\n(-)(+) = Negative', fontsize=8.5, color='darkred', alpha=0.75, weight='bold')
ax1.text(2.2, 52, 'Quadrant III\n(-)(-) = Positive', fontsize=8.5, color='darkgreen', alpha=0.75, weight='bold')
ax1.text(8.0, 52, 'Quadrant IV\n(+)(-) = Negative', fontsize=8.5, color='darkred', alpha=0.75, weight='bold')

# Plot Fitted OLS Trendline
x_vals = np.linspace(1, 11, 100)
y_vals = beta_0 + beta_1 * x_vals
ax1.plot(x_vals, y_vals, color='#e74c3c', linewidth=2.5, label=f'OLS Fit: Ŷ = {beta_0:.1f} + {beta_1:.2f}X')

# Plot Points and Vertical Residual lines
ax1.scatter(x, y, color='#2980b9', s=95, zorder=5, edgecolor='black', linewidth=1.2, label='Student Observations')
for xi, yi in zip(x, y):
    y_fit = beta_0 + beta_1 * xi
    ax1.plot([xi, xi], [yi, y_fit], color='#8e44ad', linestyle=':', linewidth=1.5)

# Annotate points
for i, st in enumerate(students):
    ax1.annotate(f"{st} ({int(x[i])}h, {int(y[i])}m)", (x[i] + 0.2, y[i] - 1.5), fontsize=8.5, weight='bold')

# Annotate prediction at X=7
y_pred_7 = beta_0 + beta_1 * 7.0
ax1.scatter([7.0], [y_pred_7], color='#f39c12', s=120, marker='*', zorder=6, label=f'Pred (7h) = {y_pred_7:.2f}')

ax1.set_title(f'1. Numerical Example: Scatter & OLS Fit (r = +{r_pearson_manual:.4f})', fontweight='bold', fontsize=11)
ax1.set_xlabel('Study Hours (X)', fontsize=10)
ax1.set_ylabel('Exam Score (Y)', fontsize=10)
ax1.set_xlim(1, 11.5)
ax1.set_ylim(45, 102)
ax1.legend(loc='lower right', fontsize=8.5)
ax1.grid(True, linestyle=':', alpha=0.5)

# ------------------------------------------------------------------------------
# Subplot 2: Spearman Rank Parity Comparison
# ------------------------------------------------------------------------------
ax2.scatter(rank_x, rank_y, color='#27ae60', s=110, zorder=5, edgecolor='black', linewidth=1.2, label='Student Ranks')
ax2.plot([0.5, 5.5], [0.5, 5.5], color='black', linestyle='--', linewidth=1.5, label='Monotonic Parity Line (y = x)')

for i, st in enumerate(students):
    d_val = int(d[i])
    ax2.annotate(f"{st}\nRx={int(rank_x[i])}, Ry={int(rank_y[i])}\n(d={d_val:+d})",
                 (rank_x[i] + 0.12, rank_y[i] - 0.25), fontsize=8.5, weight='bold')

ax2.set_title(f'2. Spearman Rank Alignment (ρ = {rho_spearman_manual:.4f})', fontweight='bold', fontsize=11)
ax2.set_xlabel('Rank of X (Study Hours)', fontsize=10)
ax2.set_ylabel('Rank of Y (Exam Score)', fontsize=10)
ax2.set_xticks(range(1, 6))
ax2.set_yticks(range(1, 6))
ax2.set_xlim(0.5, 5.8)
ax2.set_ylim(0.5, 5.8)
ax2.legend(loc='upper left', fontsize=8.5)
ax2.grid(True, linestyle=':', alpha=0.5)

plt.tight_layout()
plt.show()
```

## 2. Empirical Population Simulation ($n = 150$ Students)

Having solved the small numerical problem manually, we scale to a full sample of $n = 150$ students to observe how Covariance, Pearson correlation, and Ordinary Least Squares behave with continuous empirical distributions under random Gaussian noise $\epsilon \sim \mathcal{N}(0, 5^2)$:
$$\text{Exam Score} = 40.0 + 1.25 \times (\text{Study Hours}) + \epsilon$$

```python
np.random.seed(42)

# ==============================================================================
# 2. GENERATE BIVARIATE LINEAR RELATIONSHIP (Study Hours vs Exam Score)
# ==============================================================================
n_sim = 150
study_hours = np.random.uniform(5, 45, n_sim)
# True relationship: Exam Score = 40 + 1.25 * Hours + Gaussian Noise
noise = np.random.normal(0, 5, n_sim)
exam_scores = 40.0 + 1.25 * study_hours + noise

# Compute Covariance (Sample, ddof=1)
cov_matrix = np.cov(study_hours, exam_scores, ddof=1)
cov_xy = cov_matrix[0, 1]

# Compute Pearson Correlation Coefficient
r_pearson, p_pearson = stats.pearsonr(study_hours, exam_scores)

# Compute Spearman Rank Correlation
rho_spearman, p_spearman = stats.spearmanr(study_hours, exam_scores)

# OLS Linear Regression line parameters
beta_1_sim = cov_xy / np.var(study_hours, ddof=1)
beta_0_sim = np.mean(exam_scores) - beta_1_sim * np.mean(study_hours)

print("=" * 65)
print("      BIVARIATE ANALYSIS: STUDY HOURS VS EXAM SCORES (n=150)    ")
print("=" * 65)
print(f"Sample Size (n)               : {n_sim}")
print(f"Sample Covariance Cov(X, Y)   : {cov_xy:.2f} (hours * marks)")
print(f"Pearson Correlation (r)       : {r_pearson:.4f} (p-value: {p_pearson:.3e})")
print(f"Spearman Rank Corr (rho)      : {rho_spearman:.4f} (p-value: {p_spearman:.3e})")
print(f"Regression Trendline Equation : Y = {beta_0_sim:.2f} + {beta_1_sim:.2f} * X")
print(f"R-squared (Variance Explained): {r_pearson**2:.4f} ({r_pearson**2 * 100:.1f}%)")
print("=" * 65)
```

## 3. The Critical Exam Trap: $r \approx 0$ Does NOT Mean Independence!

Pearson's $r$ measures strictly **linear** association. If $Y = X^2$ (a perfect parabola centered at 0), $Y$ is completely deterministic from $X$, yet Pearson's $r = 0.0$!

Below we contrast:
1. **Strong Positive Linear** ($r > 0.9$)
2. **Strong Negative Linear** ($r < -0.8$)
3. **Quadratic Non-Linear Trap** ($Y = X^2$, $r \approx 0.0$ despite exact dependence)
4. **Exponential Monotonic** ($Y = e^X$, comparing Pearson $r$ vs Spearman $\rho$)

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 10))

# --- 1. Positive Linear ---
axes[0, 0].scatter(study_hours, exam_scores, color='#2980b9', alpha=0.75, edgecolor='black')
x_line = np.linspace(study_hours.min(), study_hours.max(), 100)
axes[0, 0].plot(x_line, beta_0_sim + beta_1_sim * x_line, color='#e74c3c', linewidth=2.5,
                label=f'Fit: Y={beta_0_sim:.1f}+{beta_1_sim:.2f}X')
axes[0, 0].set_title(f'1. Strong Positive Linear (r = +{r_pearson:.2f})', fontweight='bold')
axes[0, 0].set_xlabel('Study Hours')
axes[0, 0].set_ylabel('Exam Score')
axes[0, 0].legend()
axes[0, 0].grid(True, linestyle=':', alpha=0.6)

# --- 2. Negative Linear ---
x_neg = np.random.uniform(10, 50, n_sim)
y_neg = 100 - 1.8 * x_neg + np.random.normal(0, 6, n_sim)
r_neg, _ = stats.pearsonr(x_neg, y_neg)
axes[0, 1].scatter(x_neg, y_neg, color='#c0392b', alpha=0.75, edgecolor='black')
m_neg, b_neg = np.polyfit(x_neg, y_neg, 1)
x_neg_line = np.linspace(x_neg.min(), x_neg.max(), 100)
axes[0, 1].plot(x_neg_line, b_neg + m_neg * x_neg_line, color='black', linestyle='--', linewidth=2)
axes[0, 1].set_title(f'2. Strong Negative Linear (r = {r_neg:.2f})', fontweight='bold')
axes[0, 1].set_xlabel('Screen Distraction (Hours)')
axes[0, 1].set_ylabel('Performance Rating')
axes[0, 1].grid(True, linestyle=':', alpha=0.6)

# --- 3. Non-Linear Trap: Y = X^2 ---
x_quad = np.linspace(-10, 10, 200)
y_quad = x_quad**2 + np.random.normal(0, 3, 200)
r_quad, _ = stats.pearsonr(x_quad, y_quad)
axes[1, 0].scatter(x_quad, y_quad, color='#8e44ad', alpha=0.7, edgecolor='black')
axes[1, 0].set_title(f'3. Non-Linear Parabola (r = {r_quad:+.3f} ≈ 0! Strictly Dependent)', fontweight='bold')
axes[1, 0].set_xlabel('X (Centered at 0)')
axes[1, 0].set_ylabel('Y = X² + noise')
axes[1, 0].grid(True, linestyle=':', alpha=0.6)

# --- 4. Monotonic Exponential: Pearson vs Spearman ---
x_exp = np.linspace(1, 5, 100)
y_exp = np.exp(x_exp) + np.random.normal(0, 5, 100)
r_p_exp, _ = stats.pearsonr(x_exp, y_exp)
r_s_exp, _ = stats.spearmanr(x_exp, y_exp)
axes[1, 1].scatter(x_exp, y_exp, color='#16a085', alpha=0.75, edgecolor='black')
axes[1, 1].set_title(f'4. Monotonic Exponential (Pearson r={r_p_exp:.2f}, Spearman ρ={r_s_exp:.2f})', fontweight='bold')
axes[1, 1].set_xlabel('X')
axes[1, 1].set_ylabel('Y = exp(X)')
axes[1, 1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 4. Key Takeaways & Exam Reference Card

| Statistical Metric | Formula | Value Range | Scale Dependent? | Best Used When |
| :--- | :---: | :---: | :---: | :--- |
| **Covariance** | $Cov(X,Y) = \frac{\sum(X-\bar{X})(Y-\bar{Y})}{n-1}$ | $(-\infty, +\infty)$ | **Yes** (Units: $X \cdot Y$) | Determining initial joint directional movement. |
| **Pearson Correlation ($r$)** | $r = \frac{Cov(X,Y)}{s_X s_Y}$ | $[-1.0, +1.0]$ | **No** (Dimensionless) | Measuring **linear** association in continuous Gaussian data. |
| **Spearman Rank ($\rho$)** | $\rho = 1 - \frac{6\sum d_i^2}{n(n^2-1)}$ | $[-1.0, +1.0]$ | **No** (Dimensionless) | Non-linear monotonic curves, ordinal data, or data with heavy outliers. |
| **OLS Slope ($\beta_1$)** | $\beta_1 = \frac{Cov(X,Y)}{Var(X)} = r \frac{s_Y}{s_X}$ | $(-\infty, +\infty)$ | **Yes** (Units: $Y / X$) | Quantifying expected unit change in $Y$ per unit change in $X$. |
| **OLS Intercept ($\beta_0$)** | $\beta_0 = \bar{Y} - \beta_1 \bar{X}$ | $(-\infty, +\infty)$ | **Yes** (Units: $Y$) | Expected baseline value of $Y$ when $X = 0$. |
| **Variance Explained ($R^2$)** | $R^2 = r^2$ | $[0.0, 1.0]$ | **No** (Percentage) | Evaluating regression goodness of fit. |

> 💡 **Viva Golden Rule:** If an examiner asks, *"If correlation $r = 0$, does that mean variables are completely independent?"*, always answer:  
> *"No. A zero Pearson correlation only proves the absence of a **linear** association. Perfect deterministic non-linear dependencies (such as $Y = X^2$ centered at 0) can exist with $r = 0.0$."*
