```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
np.random.seed(42)
```

```python
data = np.random.normal(size=100,loc= 50.0, scale=10.0) # loc is the mean, scale is the standard deviation, size is the number of samples
data_woth_dirt = data.copy()
data_woth_dirt[0] = 1000000  # introduce an outlier
print("mean for prestine data:", np.mean(data))
print("mean for data with outlier:", np.mean(data_woth_dirt))
print("median for prestine data:", np.median(data))
print("median for data with outlier:", np.median(data_woth_dirt))
# trimmed mean by 10%
print("trimmed mean for prestine data:", stats.trim_mean(data, 0.1))
print("trimmed mean for data with outlier:", stats.trim_mean(data_woth_dirt, 0.1))
```

```python
# empirical verification of Bessel's correction via monte carlo simulation
chips = np.random.normal(size=100000,loc= 100.0, scale=15.0)
true_variance = np.var(chips)
sample_variance = np.var(chips, ddof=1)  # unbiased estimator of variance : ddof -> delta degrees of freedom
sample_size = 10
num_trials = 1000
biased_variance = []
unbiased_variance = []
for _ in range(num_trials):
    sample = np.random.choice(chips, size=sample_size, replace = False)
    biased_variance.append(np.var(sample,ddof=0)) # N
    unbiased_variance.append(np.var(sample,ddof=1)) # N-1
print("true variance:", true_variance)
print(f"Unbiased Estimation (N-1) : {np.mean(unbiased_variance):.4f}")
print(f"Biased Estimation (N)     : {np.mean(biased_variance):.4f}")
```

#### Welford's Algorithm (Numerical Stable streaming estimator)
Welford's algorithm updates mean $M_k$ and sum of squared differences $S_k$ incrementally:
$$M_k = M_{k-1} + \frac{x_k - M_{k-1}}{k}$$
$$S_k = S_{k-1} + (x_k - M_{k-1})(x_k - M_k)$$
