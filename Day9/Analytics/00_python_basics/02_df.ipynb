```python
import pandas as pd
import numpy as np
```

```python
# Constructing a DataFrame from a dictionary of lists
df_cricket = pd.DataFrame({
    "name": ["Shubman Gill", "Deepti Sharma", "Mohammed Siraj", "Virat Kohli",
             "Smriti Mandhana", "Jasprit Bumrah", "Richa Ghosh", "Hardik Pandya"],
    "age": [25, 27, 31, 36, 28, 31, 21, 31],
    "gender": ["Male", "Female", "Male", "Male", "Female", "Male", "Female", "Male"]
})

print("--- In-Memory DataFrame ---")
print(df_cricket)

# Extracting a single column returns a Series
age_series = df_cricket["age"]
print("\nSingle Column Type:", type(age_series))
print("Summary Statistics:\n", age_series.describe())
print("_" * 60)

# info 
print("\nDataFrame Info:")
print(df_cricket.info())    
```

```python
# Sample Wide Dataset: Player statistics
df_wide = pd.DataFrame({
    'player': ['Kohli', 'Rohit', 'Gill'],
    'fifties': [220, 180, 45],
    'hundreds': [80, 31, 6]
})

print("--- Original Wide Format ---")
print(df_wide)
print("_" * 60)
# Converting from wide to long format (unpivot) => melt
df_long = pd.melt(df_wide, id_vars=['player'], value_vars=['fifties', 'hundreds'],
                  var_name='ScoreTypes', value_name='count')
print("\n--- Long Format ---")
print(df_long)
df_wide_restored = df_long.pivot(index='player', columns='ScoreTypes', values='count').reset_index()
print("\n--- Restored Wide Format ---")
print(df_wide_restored)
```

```python
df_sales = pd.DataFrame({
    'region': ['North', 'North', 'South', 'South', 'North'],
    'product': ['A', 'B', 'A', 'B', 'A'],
    'revenue': [100, 150, 200, 120, 130]
})

# Aggregate revenue across duplicate entries
pt = df_sales.pivot_table(index='region', columns='product', values='revenue', aggfunc='sum', fill_value=0)
print("--- Pivot Table (Sum Revenue) ---")
print(pt)
```

```python
df_multi = pd.DataFrame({
    'Q1': [10, 20],
    'Q2': [15, 25]
}, index=['Company_A', 'Company_B'])
print("--- Multi-Index DataFrame ---")
print(df_multi)

# accessing data in the multi-index DataFrame
print("\nAccessing Q1 column:")
print(df_multi['Q1'])

print("\nAccessing Q2 column:")
print(df_multi['Q2'])

print("\nAccessing data for Company_A:")
print(df_multi.loc['Company_A'])

#iloc for positional indexing
print("\nAccessing first row using iloc:")
print(df_multi.iloc[0])

print("\nAccessing first column using iloc:")
print(df_multi.iloc[:, 0])
```

```python
df_multi = pd.DataFrame({
    'Q1': [10, 20],
    'Q2': [15, 25]
}, index=['Company_A', 'Company_B'])
print("--- Multi-Index DataFrame ---")
print(df_multi)

# Stack columns into a MultiIndex Series
stacked = df_multi.stack()
print("--- Stacked Series ---")
print(stacked)

# Unstack the MultiIndex Series back to a DataFrame
unstacked = stacked.unstack()
print("--- Unstacked DataFrame ---")
print(unstacked)
```

```python
df_nested = pd.DataFrame({
    'project': ['Alpha', 'Beta'],
    'tags': [['Python', 'C++'], ['Pandas', 'SQL', 'Docker']]
})

print("--- Nested Lists ---")
print(df_nested)

df_exploded = df_nested.explode('tags')
print("\n--- Exploded DataFrame ---")
print(df_exploded)
```
